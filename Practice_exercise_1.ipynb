John is an investor with Apple stocks. Each Apple stock is currently valued at $135. He bought 20 stocks at $60 each, then acquired an additional 10 at $145 each. He sold 8 stocks at $120 each. Now, John is evaluating the financial merits of selling his remaining stocks and is curious about the potential tax implications.

```python
stock_value_now = 135 #present price
stocks_bought_first = 20
price_first = 60 #price when he bought 20 stocks
stocks_bought_second = 10
price_second = 145 #price when bought 10 stocks later
stocks_sold = 8
price_sold = 120 #he sold 8 stocks at the rate of 120
```

Calculate and display the total value of the current Apple stocks in John’s portfolio.

```python
total_stocks = stocks_bought_first + stocks_bought_second - stocks_sold
total_value = stock_value_now * total_stocks
print(f"The total value of current Apple stocks in John's portfolio: ${total_value:.2f}")
```

Calculate and display the average purchase price of John’s Apple stocks.

```python
total_stocks_purchase_price = stocks_bought_first*price_first + stocks_bought_second*price_second 
combined_stocks = stocks_bought_first + stocks_bought_second    
average_stocks_purchase_price = total_stocks_purchase_price/combined_stocks
print(f"The average purchase price of John's Apple stocks: ${average_stocks_purchase_price:.2f}")
```

Calculate and display the potential total capital gain if John sells all his Apple stocks now.

```python
remaining_stocks = stocks_bought_first - stocks_sold
pnl_remaining_stocks = remaining_stocks*(stock_value_now-price_first)
pnl_second_stocks = stocks_bought_second*(stock_value_now-price_second)
pnl_sold_stocks = (price_sold-price_first)*stocks_sold
total_capital_gain = pnl_remaining_stocks+pnl_second_stocks+pnl_sold_stocks
print(f"The potential total capital gain if John sells his Apple stocks now: ${total_capital_gain:.2f}")
```

John holds dual citizenship in France and Belgium. In France, the tax rate on capital gains is 30%. In Belgium, capital gains are tax-exempt up to $500, and a 40% tax rate applies beyond this threshold. Calculate and display the taxes John would incur on his capital gains in France and Belgium.

```python
tax_rate_france = 0.3
taxes_incur_france = total_capital_gain*tax_rate_france
print(f"The taxes John would incur on his capital gains in France: ${taxes_incur_france:.2F}")
tax_rate_belgium = 0.4
tax_exempt = 500
tax_incur_belgium = (total_capital_gain-tax_exempt)*tax_rate_belgium
print(f"The taxes John would incur on his capital gains in Belgium: ${tax_incur_belgium:.2F} ")
```
