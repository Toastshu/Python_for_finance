```python
initial_wallet = 1000
interest_rate = 0.05

years = 0
current_wallet = initial_wallet

while current_wallet < 2*initial_wallet:
    current_wallet = current_wallet * ( 1 + interest_rate)    
    years += 1
    print("My investment is worth", current_wallet, "on year", years)
```

```python
#Variables
saving_goal = 16000
monthly_savings = 400
month_of_raise = 4
raise_amount = 200

#Counter Initialization
months_required = 0
total_savings= 0

#Loop until you reach your goal
while total_savings < saving_goal:
    # check whether raise is in effect or not
    if months_required == month_of_raise:
        monthly_savings += raise_amount
        print("Yay! I got a raise...")
        
    #updated total savings
    total_savings += monthly_savings
    
    # increment the months counter
    months_required += 1
    print(months_required, total_savings)
    
print(f"You will need {months_required} months to reach your savings goal of {saving_goal}.")
```
